## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Recreate the label the same way ML-01's pipeline does
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

declining_rate = df["is_declining_label"].mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).mean()

print(f"Rows: {len(df)}, Declining rate: {declining_rate:.3f}")
print(f"Stale + visible pages: {stale_visible:.3f}")
print("From ML-01: baseline rule Precision@50 = 0.240 vs random forest Precision@50 = 0.740 (~3.1x better)")
```
